```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('../data/processed/analysis_ready_data.csv')
df.head()
```

```python
# Summary statistics
df.describe()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Distribution of Total Deaths
plt.figure(figsize=(10, 6))
plt.hist(df['Total_Deaths'], bins=15, edgecolor='black')
plt.xlabel('Total Deaths per Event')
plt.ylabel('Frequency')
plt.title('Distribution of Mortality Event Severity')
plt.show()
```

```python
# Correlation matrix
variables = ['Total_Deaths', 'GiZScore', 'Port_Distance_km', 
             'Density_Value', 'Risk_Score']
corr = df[variables].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='RdBu_r', center=0, fmt='.3f')
plt.title('Correlation Matrix')
plt.show()
```

```python
# Temporal trend
yearly = df.groupby('Year')['Total_Deaths'].sum().reset_index()

plt.figure(figsize=(12, 5))
plt.bar(yearly['Year'], yearly['Total_Deaths'], alpha=0.7)
plt.plot(yearly['Year'], yearly['Total_Deaths'], 'r-o', linewidth=2)
plt.xlabel('Year')
plt.ylabel('Total Deaths')
plt.title('Annual Dolphin Mortality (2000-2024)')
plt.show()
```

```python
# Cause of death
causes = {
    'Bycatch': df['Entanglement_Bycatch'].sum(),
    'Boat Strikes': df['Boat_Strikes'].sum(),
    'Other': df['Other_Causes'].sum()
}

plt.figure(figsize=(8, 6))
plt.pie(causes.values(), labels=causes.keys(), autopct='%1.1f%%',
        startangle=90, wedgeprops={'edgecolor': 'black'})
plt.title('Cause of Death Distribution')
plt.show()
```
